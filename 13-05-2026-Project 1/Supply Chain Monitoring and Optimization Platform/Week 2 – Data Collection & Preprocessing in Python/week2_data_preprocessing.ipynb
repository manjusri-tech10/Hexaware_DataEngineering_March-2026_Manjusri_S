```python
import pandas as pd
import numpy as np
import requests
```

```python
try:
    data = requests.get("https://api.sampledata.com/orders").json()
    df = pd.DataFrame(data)
except:
    # Simulate data locally
    df = pd.DataFrame({
        'order_id': [1, 2, 3, 4, 5],
        'supplier_id': ['S1', 'S2', 'S1', 'S3', 'S2'],
        'item': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam'],
        'quantity': [10, 50, 30, 5, 20],
        'delivery_date': ['2024-01-01', '2024-01-15', None, '2023-12-20', '2024-02-01'],
        'status': ['Delivered', 'Pending', 'Delayed', 'Delivered', None]
    })
```

```python
# Clean Data
df.dropna(subset=['delivery_date'], inplace=True)
df['status'].fillna('Unknown', inplace=True)
```

```python
# Format Dates
df['delivery_date'] = pd.to_datetime(df['delivery_date'])
```

```python
# Calculate Delay Days
df['delay_days'] = (pd.Timestamp.today() - df['delivery_date']).dt.days
```

```python
# Flag Delayed Orders using NumPy
df['is_delayed'] = np.where(df['delay_days'] > 0, 1, 0)
```

```python
# Basic Stock Calculations using NumPy
quantities = np.array(df['quantity'])
print(f"Total Stock Ordered: {np.sum(quantities)}")
print(f"Average Quantity: {np.mean(quantities):.2f}")
print(f"Max Order: {np.max(quantities)}, Min Order: {np.min(quantities)}")
```

```python
# Display Output
print(df[['order_id', 'supplier_id', 'delay_days', 'is_delayed']].head())
```

```python
df.to_csv('processed_orders.csv', index=False)
print("Cleaned data saved to processed_orders.csv")
```
