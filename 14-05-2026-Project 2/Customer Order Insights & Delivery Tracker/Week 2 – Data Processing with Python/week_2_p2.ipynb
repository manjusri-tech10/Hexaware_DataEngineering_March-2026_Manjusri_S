```python
import pandas as pd
import numpy as np

# Load Data
df = pd.read_csv('orders.csv')
```

```python
# Clean missing fields
df['delivery_date'] = pd.to_datetime(df['delivery_date'], errors='coerce')
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
df['issue'] = df['issue'].fillna('None')
```

```python
# Calculate delay days
df['delay_days'] = (df['delivery_date'] - df['order_date']).dt.days
```

```python
# Flag delayed orders (more than 5 days)
df['delayed'] = np.where(df['delay_days'] > 5, 1, 0)
```

```python
# Top delayed customers
print("=== Top Delayed Customers ===")
print(df.groupby('customer_id')['delayed'].sum().sort_values(ascending=False))
```

```python
# Most common delivery issues
print("\n=== Most Common Delivery Issues ===")
print(df[df['issue'] != 'None']['issue'].value_counts())
```

```python
print("\n=== Full Data Summary ===")
print(df[['order_id','customer_id','delay_days','delayed','issue']])
```
