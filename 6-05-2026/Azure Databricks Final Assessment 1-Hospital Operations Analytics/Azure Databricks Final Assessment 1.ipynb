Part 8 — Merge / Upsert

```python
# 71
visits_df.write.format("delta").mode("overwrite").save(f"{DELTA_PATH}/target_visits")
```

```python
# 72
target_dt = DeltaTable.forPath(spark, f"{DELTA_PATH}/target_visits")
target_dt.toDF().show()
```

```python
# 73
daily_visits_data = [
(26,1002,201,"2024-03-14","Completed",2),
(27,1005,210,"2024-03-14","Completed",1),
(11,1011,205,"2024-03-06","Completed",3),
(20,1020,206,"2024-03-10","Completed",2),
(28,1018,203,"2024-03-15","Pending",2)
]
daily_visits_columns = ["visit_id","patient_id","doctor_id","visit_date","visit_status","tests_conducted"]
daily_visits_df = spark.createDataFrame(daily_visits_data, daily_visits_columns)
daily_visits_df.createOrReplaceTempView("daily_visits")
```

```python
# 74
from pyspark.sql.functions import to_date

daily_visits_df = daily_visits_df.withColumn("visit_date", to_date("visit_date", "yyyy-MM-dd"))

target_dt.alias("target").merge(
    daily_visits_df.alias("source"),
    "target.visit_id = source.visit_id"
).whenMatchedUpdate(set={
    "visit_status": "source.visit_status",
    "visit_date":   "source.visit_date",
    "tests_conducted": "source.tests_conducted"
}).whenNotMatchedInsert(values={
    "visit_id":        "source.visit_id",
    "patient_id":      "source.patient_id",
    "doctor_id":       "source.doctor_id",
    "visit_date":      "source.visit_date",
    "visit_status":    "source.visit_status",
    "tests_conducted": "source.tests_conducted",
    "test_cost":       "source.tests_conducted * 500",
    "visit_priority":  "'Low'"
}).execute()
```
