Exercise 1 — Sales Revenue Summary Job

```python
from pyspark.sql.functions import col, sum, count

data = [
    (1, "Aarav Sharma", "Hyderabad", "Laptop", 1, 75000),
    (2, "Priya Reddy", "Bengaluru", "Mouse", 4, 600),
    (3, "Rohan Mehta", "Mumbai", "Monitor", 2, 12000),
    (4, "Sneha Iyer", "Chennai", "Keyboard", 3, 1500),
    (5, "Kiran Patel", "Hyderabad", "Laptop", 1, 75000),
    (6, "Ananya Das", "Kolkata", "Tablet", 2, 30000),
    (7, "Vikram Singh", "Delhi", "Mouse", 6, 600),
    (8, "Meera Nair", "Bengaluru", "Monitor", 1, 12000),
    (9, "Rahul Verma", "Mumbai", "Laptop", 1, 75000),
    (10, "Pooja Shah", "Hyderabad", "Keyboard", 5, 1500)
]

columns = ["sale_id", "customer_name", "city", "product", "quantity", "unit_price"]
df = spark.createDataFrame(data, columns)
display(df)
```

```python
#1
df = df.withColumn("revenue", col("quantity") * col("unit_price"))

```

```python
#2
df.groupBy("city").agg(sum("revenue").alias("total_revenue")).show()

```

```python
#3
df.groupBy("city").agg(sum("quantity").alias("total_quantity")).show()

```

```python
#4
df.groupBy("city").agg(sum("revenue").alias("total_revenue")).orderBy(col("total_revenue").desc()).show()
```

Exercise 2 — Student Performance Job

```python
from pyspark.sql.functions import col, avg, when

data = [
    (101, "Arjun Kumar", "Python", 86, 92),
    (102, "Neha Gupta", "Python", 74, 80),
    (103, "Rahul Nair", "Data Engineering", 88, 85),
    (104, "Priya Sharma", "AI", 91, 95),
    (105, "Siddharth Rao", "Data Engineering", 67, 72),
    (106, "Ayesha Khan", "AI", 79, 88),
    (107, "Kavya Reddy", "Python", 94, 96),
    (108, "Manish Yadav", "Data Engineering", 58, 69),
    (109, "Divya Menon", "AI", 83, 90),
    (110, "Farhan Ali", "Python", 62, 75)
]

columns = ["student_id", "student_name", "course", "marks", "attendance"]
df = spark.createDataFrame(data, columns)
display(df)
```

```python
#1
df = df.withColumn("grade", when(col("marks") >= 90, "A").when(col("marks") >= 75, "B").when(col("marks") >= 60, "C").otherwise("Needs Improvement"))
```

```python

#2
df = df.withColumn("certification_status", when((col("marks") >= 75) & (col("attendance") >= 80), "Eligible").otherwise("Not Eligible"))

```

```python
#3
df.groupBy("course").agg(avg("marks").alias("avg_marks")).show()

```

```python
#4
df.filter(col("certification_status") == "Eligible").groupBy("course").count().show()
```

Exercise 3 — Employee Department Ranking
Job

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import col, rank, dense_rank, avg, when

data = [
    (1, "Ravi Kumar", "Engineering", "Hyderabad", 92000),
    (2, "Sneha Reddy", "Engineering", "Bengaluru", 97000),
    (3, "Amit Shah", "Finance", "Mumbai", 88000),
    (4, "Pooja Nair", "HR", "Chennai", 69000),
    (5, "Nikhil Verma", "Engineering", "Delhi", 85000),
    (6, "Meera Iyer", "Finance", "Hyderabad", 91000),
    (7, "Karan Malhotra", "Marketing", "Mumbai", 76000),
    (8, "Anjali Rao", "HR", "Bengaluru", 72000),
    (9, "Vivek Joshi", "Finance", "Delhi", 94000),
    (10, "Fatima Khan", "Marketing", "Chennai", 81000),
    (11, "Aditya Menon", "Engineering", "Mumbai", 99000),
    (12, "Lakshmi Das", "Finance", "Bengaluru", 87000)
]

columns = ["emp_id", "emp_name", "department", "city", "salary"]
df = spark.createDataFrame(data, columns)
display(df)
```

```python
#1
window_dept = Window.partitionBy("department").orderBy(col("salary").desc())
df = df.withColumn("rank", rank().over(window_dept))

```

```python
#2
df = df.withColumn("salary_level", when(col("salary") >= 95000, "Very High").when(col("salary") >= 85000, "High").otherwise("Standard"))
```

```python

#3
df.filter(col("rank") <= 2).show()

```

```python
#4
df.groupBy("department").agg(avg("salary").alias("avg_salary")).show()
```
