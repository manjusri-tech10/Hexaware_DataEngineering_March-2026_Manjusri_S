Final Mini Capstone

```python
%sql
CREATE TABLE employee_salary_analysis AS
SELECT
    e.emp_name,
    e.city,
    e.department,
    e.salary,
    e.experience,
    CASE
        WHEN e.salary >= 95000 THEN 'High'
        WHEN e.salary >= 80000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_category,
    d.avg_sal AS department_avg_salary
FROM employees e JOIN (SELECT department, AVG(salary) AS avg_sal FROM employees GROUP BY department) d
  ON e.department = d.department;
```
